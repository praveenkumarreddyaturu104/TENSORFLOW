```python
#SET SEED
tf.random.set_seed(42)

#creating model
model_1=tf.keras.Sequential([
    tf.keras.layers.Dense(100,activation="relu"),
    tf.keras.layers.Dense(10,activation="relu"),
    tf.keras.layers.Dense(1,activation="sigmoid")
])

#compile the model
model_1.compile(loss=tf.keras.losses.BinaryCrossentropy(),
                optimizer=tf.keras.optimizers.Adam(),
                metrics=["accuracy"]
                )

#fit the model
h=model_1.fit(x,y,epochs=200)
```

```python
plt.plot(h.history['accuracy'])
plt.title("Model Accuracy over Epochs")
plt.show()
```

```python
y_pred=model_1.predict(x)
plt.figure(figsize=(10,10))
plt.scatter(x[:,0],x[:,1], y_pred.flatten(), c=y, cmap=plt.cm.RdYlBu)
plt.show
plt.scatter(x[:,0],x[:,1],c=y,s=40,cmap=plt.cm.RdYlBu)
plt.legend()
```

```python
import numpy as np
```

```python
def plot_decision_boundary(model_1,x,y):
  """
  plot the decion boundary created by our model predicting on x
  """
  #define the axis bondary created by a  model predicting on x
  x_min,x_max=x[:,0].min()-0.1,x[:,0].max()+0.1
  y_min,y_max=x[:,1].min()-0.1,x[:,1].max()+0.1
  xx,yy=np.meshgrid(np.linspace(x_min,x_max,100),np.linspace(y_min,y_max,100))
  #create x value (we're going to make predictions on these)
  x_in=np.c_[xx.ravel(),yy.ravel()]#stacking 2d arrays to together

  #make predictions
  y_prd=model_1.predict(x_in)

  #check for multi-class
  if len(y_prd[0])>1:
     print("during multiclass classifiction")
    #we have to reshape our prediction to get them eady for plotting
     y_prd=np.argmax(y_prd,axis=1).reshaoe(xx.shape)
  else:
    print("doing binary classifiction")
    y_prd=np.round(y_prd).reshape(xx.shape)

  #plot the decision boundray
  plt.contourf(xx,yy,y_prd,cmap=plt.cm.RdYlGn,aplha=0.7)
  plt.scatter(x[:,0],x[:,1],c=y,s=40,cmap=plt.cm.RdYlGn)
  plt.xlim(xx.min(),xx.max())
  plt.ylim(yy.min(),yy.max())
```

```python
#check our the predictions
plot_decision_boundary(model_1,x=x,y=y)
```

```python
#spliting into train and test
x_train,y_train=x[:800],y[:800]
x_test,y_test=x[800:],y[800:]
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
#lets recreate a model tpo fit on the training data and evaluate on the data
#set random seed
tf.random.set_seed(42)

#creating model
model_2=tf.keras.Sequential([
    tf.keras.layers.Dense(40,activation="relu"),
    tf.keras.layers.Dense(40,activation="relu"),
    tf.keras.layers.Dense(1,activation="sigmoid")
])

#compiling the model
model_2.compile(
    loss=tf.keras.losses.binary_crossentropy,
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
    metrics=["accuracy"]
)

#fitting model
h=model_2.fit(x_train,y_train,epochs=50)
```

```python
model_2.summary()
```

```python
y_pred=model_2.predict(x_test)
```

```python
model_2.evaluate(x_test,y_test)
```

```python
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.title("TRAIN")
plot_decision_boundary(model_2,x=x_train,y=y_train)
plt.subplot(1,2,2)
plt.title("TEST")
plot_decision_boundary(model_2,x_test,y_test)
```

```python
pd.DataFrame(h.history).plot()
plt.title("MODEL_2")
```

```python
#ser random seed
tf.random.set_seed(42)

#creating model_3
model_3=tf.keras.Sequential([
    tf.keras.layers.Dense(4,activation="relu"),
    tf.keras.layers.Dense(4,activation="relu"),
    tf.keras.layers.Dense(1,activation="sigmoid")
])

#compile modek_3
model_3.compile(loss="binary_crossentropy",
                optimizer="Adam",
                metrics=["accuracy"])
#creating learning rate callback
lr_scheduler=tf.keras.callbacks.LearningRateScheduler(lambda epoch:1e-4*10**(epoch/20))

#fit the model(passing lr_scheduler callback)
h_3=model_3.fit(x_train,y_train,epochs=100,callbacks=[lr_scheduler])
```

```python
#ploting the graph
pd.DataFrame(h_3.history).plot()
```

```python
#plot the learning rate vs loss
lrs=1e-4*(10**(tf.range(100)/20))
lrs
```
