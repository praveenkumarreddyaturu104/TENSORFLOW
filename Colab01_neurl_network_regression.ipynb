```python
#introduction to regression with neural networks
#neural network in trensorflow
#import tnesor flow
import tensorflow as tf
print(tf.__version__)
```

```python
import numpy as np #for array feature
import matplotlib.pyplot as plt #for visulization freature
#creating sample data
x=np.array([-4.0,-2.0,1.0,2.5,3.0,9.0,8.0,-7.0])
y=np.array([4.0,5.0,9.0,7.0,-9.0,7.0,2.0,3.0])
plt.scatter(x,y)
```

```python
#checking input and output array size to simulate actual neural networks
house_info=tf.constant(['bathroom','bedroom','garage'])
house_prise=tf.constant([94000])
house_info,house_prise
```

```python
#checking shape of each sample data
input_shape=x.shape
output_shape=y.shape
input_shape,output_shape
```

```python
#checking the dimensions of x sample
x.ndim
```

```python
#converting into tensorflow cnstants
X=tf.constant(x)
Y=tf.constant(y)
X,Y
```

```python
#steps in modeling with tensorflow
#craeting a modedl-deine the input and output layers,as well as the hidden layers
#compile a model
#fitting moddel
#set seed
tf.random.set_seed(42)

#creating model
model=tf.keras.Sequential([
    tf.keras.layers.Dense(1)
])

#compile the model
model.compile(loss=tf.keras.losses.mae,
              optimizer=tf.keras.optimizers.SGD(),
              metrics=["mae"])

# Reshape X to be a 2D tensor (samples, features)
X_reshaped = tf.expand_dims(X, axis=-1)

#fit the model
model.fit(X_reshaped,y,epochs=10)
```

```python
#predicting to check the model
p=model.predict(tf.constant([8.5]))
p
```

```python
#imporving model
#hint small model---->larger model


#creating model
model_3=tf.keras.Sequential([
    tf.keras.layers.Dense(100,activation="relu"),
    tf.keras.layers.Dense(100,activation="relu"),
    tf.keras.layers.Dense(100,activation="relu"),
    tf.keras.layers.Dense(1)
])

#compile model
model_3.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                metrics=["mae"]
                )
#fit
model_3.fit(X_reshaped,y,epochs=100)
```

```python
def plot_predictions(train_data=x_train,
                     label_data=y_train,
                     test_data=x_test,
                     test_lables=y_test,
                    predicisctions=y_pred):
  #plt figure size setting
  plt.figure(figsize=(10,7))
  #ploting training data in orange
  plt.scatter(train_data,label_data,c="orange",label="traing data")
  #ploting test data in blue
  plt.scatter(test_data,test_lables,c="blue",label="test data")
  #ploting model's prediction in red
  plt.scatter(test_data,predicisctions,c="green",label="predictions")
  #ploting graph
  plt.legend();
```

```python
#importing required libreries
import tensorflow as tf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#read in the insurance
insurance=pd.read_csv("https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/refs/heads/master/insurance.csv")
insurance
```

```python
insurance.describe()
```

```python
insurance_one_hot=pd.get_dummies(insurance)
insurance_one_hot
```

```python
#create x,y labels
x=insurance_one_hot.drop("charges",axis=1)
y=insurance_one_hot["charges"]
```

```python
#creating training and test sets
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
len(x_train),len(x_test),len(y_train),len(y_test)
```

```python
#build a neural network
tf.random.set_seed(42)

#craete model
model_4=tf.keras.Sequential([
    tf.keras.layers.Dense(100,activation="relu"),
    tf.keras.layers.Dense(50,activation="relu"),
    tf.keras.layers.Dense(25,activation="relu"),
    tf.keras.layers.Dense(10,activation="relu"),
    tf.keras.layers.Dense(5,activation="relu"),
    tf.keras.layers.Dense(1)
])

#model compiling
model_4.compile(loss=tf.keras.losses.mae,
                optimizer=tf.keras.optimizers.Adam(),
                metrics=["mae"])

#model fit
h=model_4.fit(x_train,y_train,epochs=200)
```

```python
#model testing
y_pred=model_4.predict(x_test)
y_pred
```

```python
model_4.evaluate(x_test,y_test)
```

```python
#plot history
pd.DataFrame(h.history).plot()
plt.ylabel("LOSS")
plt.xlabel("EPOCHS")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import numpy as np
```

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import MinMaxScaler,OneHotEncoder

#create a column transformer
ct=make_column_transformer(
    (MinMaxScaler(),["age","bmi","children"]),#turn all values in these columns btw 0 and 1
    (OneHotEncoder(handle_unknown="ignore"),["sex","smoker","region"])
)
#creating x & y
x=insurance.drop("charges",axis=1)
y=insurance["charges"]

#build our train and test sets
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

#fit the column transformer to our training data
ct.fit(x_train)

#transform training and test data with normalization MinMaHotEncoder
x_train_normal=ct.transform(x_train)
x_test_normal=ct.transform(x_test)
```
